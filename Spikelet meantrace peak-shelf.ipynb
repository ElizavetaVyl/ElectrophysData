# Spikelet meantrace (peak / shelf auto)

Отдельный пайплайн только по **усреднённому spikelet** (meantrace).

1. Выберите папку с `.abf`
2. В списке **кликните файлы базового периода** (baseline)
3. Считаются amp / ratio / delay; нормализация: `value / mean(baseline)` по направлению

Выход в `Spikelet_meantrace_peak_shelf/`:
- Excel (сырые + `*_norm` + `is_baseline`)
- QC PNG на файл
- over-time raw + over-time **norm**

```python
import importlib
import spikelet_meantrace_folder as sm
importlib.reload(sm)
from spikelet_meantrace_folder import (
    analyze_folder,
    _ask_folder,
    list_abf_files,
    _ask_baseline_files,
)

print("CONFIG:")
print("  WINDOW_MS", sm.WINDOW_MS)
print("  PEAK_PROMINENCE_MV", sm.PEAK_PROMINENCE_MV)
print("  SHELF_DROP_FRAC", sm.SHELF_DROP_FRAC)
print("  SPIKELET_ONSET_FRAC", sm.SPIKELET_ONSET_FRAC)
print("  SMOOTH_MS", sm.SMOOTH_MS)
```

```python
FOLDER_PATH = None  # or r"/path/to/abf/folder"
# Optional: skip the clickable picker and hardcode baseline names:
BASELINE_FILES = None  # e.g. ["rec_0041.abf", "rec_0042.abf"]

folder_path = FOLDER_PATH or _ask_folder()
print("Folder:", folder_path)

if not folder_path:
    print("No folder selected.")
else:
    files = list_abf_files(folder_path)
    print(f"\nFiles in folder ({len(files)}):")
    for i, fn in enumerate(files, 1):
        print(f"  {i:3d}. {fn}")

    if BASELINE_FILES is None:
        print("\n→ Click baseline files in the dialog, then OK")
        baseline_files = _ask_baseline_files(files)
    else:
        baseline_files = list(BASELINE_FILES)

    print("Baseline:", baseline_files or "(none — no normalization)")
    result = analyze_folder(
        folder_path,
        save_qc=True,
        baseline_files=baseline_files or None,
        ask_baseline=False,  # already asked above
    )
    print(result)
```
